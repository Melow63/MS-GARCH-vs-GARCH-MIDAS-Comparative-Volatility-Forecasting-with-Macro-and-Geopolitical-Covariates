```python
import datetime as dt
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
import pandas as pd
import numpy as np
from scipy import stats
from arch import arch_model
```

```python
data = pd.read_csv("../data/processed/merged_data.csv", index_col=0, parse_dates=True)

am = arch_model(data["log_return"], vol="GARCH", p=1, q=1, rescale = True)
res = am.fit(update_freq=5)
print(res.summary())

returns_scaled = data["log_return"] * 100

am = arch_model(returns_scaled, vol="GARCH", p=1, q=1, rescale=False)
res = am.fit(disp="off")
print(f"BIC = {res.bic:.3f}")
print(f"LLH = {res.loglikelihood:.3f}")
```
